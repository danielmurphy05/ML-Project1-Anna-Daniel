```python
import numpy as np

# Calculer la matrice de corrélation
correlation_matrix = data_clean.corr()

# Définir un seuil pour les corrélations fortes
threshold = 0.99

# Extraire les paires de colonnes fortement corrélées
strong_correlations = []
for col1 in correlation_matrix.columns:
    for col2 in correlation_matrix.columns:
        if col1 != col2 and abs(correlation_matrix[col1][col2]) > threshold:
            strong_correlations.append((col1, col2, correlation_matrix[col1][col2]))

# Supprimer les doublons (par exemple, (A, B) et (B, A))
strong_correlations = list(set(tuple(sorted(pair[:2])) + (pair[2],) for pair in strong_correlations))

# Afficher les colonnes fortement corrélées
if strong_correlations:
    print("Colonnes fortement corrélées (seuil > 0.8) :")
    for col1, col2, corr in strong_correlations:
        print(f"{col1} et {col2} : corrélation = {corr:.2f}")
else:
    print("Aucune colonne fortement corrélée trouvée.")
```

```python
# Supprimer une des deux colonnes fortement corrélées
for col1, col2, corr in strong_correlations:
    if col1 in data_clean.columns and col2 in data_clean.columns:
        data_clean = data_clean.drop(columns=[col2])
        print(f"Colonne supprimée : {col2}")
```

### Linear regression using gradient descent

```python
def compute_mse(y, tx, w):
    e = y - tx @ w
    return np.mean(e**2) / 2

def compute_gradient(y, tx, w):
    e = y - tx @ w
    return -(tx.T @ e) / len(y)


def gradient_descent(y, tx, initial_w, max_iters, gamma):
    w = initial_w.copy()

    for _ in range(max_iters):
        gradient = compute_gradient(y, tx, w)
        w = w - gamma * gradient

    return w, compute_mse(y, tx, w)
```
